### 1. Membaca file real_estate.csv

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('real_estate.csv')
print(df.head())
```

### 2. Menampilkan variable dalam dataset

```python
print("Nama kolom:", list(df.columns))


X = df.iloc[:, 1:-1].values
y = df.iloc[:, -1].values

X1 = df['X1 transaction date'].values
X2 = df['X2 house age'].values
X3 = df['X3 distance to the nearest MRT station'].values
X4 = df['X4 number of convenience stores'].values
X5 = df['X5 latitude'].values
X6 = df['X6 longitude'].values
Y = df['Y house price of unit area'].values

print("Variabel X1 (transaction date), 5 data pertama:\n", X1[:5])
print("Variabel X2 (house age), 5 data pertama:\n", X2[:5])
print("Variabel X3 (distance to the nearest MRT station), 5 data pertama:\n", X3[:5])
print("Variabel X4 (number of convenience stores), 5 data pertama:\n", X4[:5])
print("Variabel X5 (latitude), 5 data pertama:\n", X5[:5])
print("Variabel X6 (longitude), 5 data pertama:\n", X6[:5])
print("Variabel Y (house price of unit area), 5 data pertama:\n", Y[:5])
print("\nDimensi X:", X.shape)
print("Dimensi y:", y.shape)
```

### 3. Menampilkan data yang di-split train: 80% & test: 20%

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

print("X_train:\n", X_train)
print("y_train:\n", y_train)
print("X_test:\n", X_test)
print("y_test:\n", y_test)
print("Jumlah data train:", len(X_train), "| Jumlah data test:", len(X_test))
```

### 4. Memanggil LinearRegression()

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

### 5. Menampilkan hasil predict dari data test

```python
y_pred = regressor.predict(X_test)
y_pred
```

### 6. Menampilkan Grafik Nilai Penjualan berdasarkan Biaya Promosi

```python
plt.scatter(y_test, y_pred, color='red')
plt.title('Grafik Nilai Penjualan berdasarkan Biaya Promosi (Harga Rumah)')
plt.xlabel('Nilai Aktual')
plt.ylabel('Nilai Prediksi')
plt.show()
```

### 7. Menampilkan Garis hasil prediksi dari machine learning pada Biaya Promosi terhadap penjualan (train set dan test set)

```python
plt.scatter(y_train, regressor.predict(X_train), color='red')
plt.plot([y_train.min(), y_train.max()], [y_train.min(), y_train.max()], color='blue')
plt.title('Nilai Penjualan vs Biaya Promosi / Harga Rumah (Train set)')
plt.xlabel('Nilai Aktual')
plt.ylabel('Nilai Prediksi')
plt.show()
```

```python
plt.scatter(y_test, y_pred, color='red')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], color='blue')
plt.title('Nilai Penjualan vs Biaya Promosi / Harga Rumah (Test set)')
plt.xlabel('Nilai Aktual')
plt.ylabel('Nilai Prediksi')
plt.show()
```

## 2. Hitunglah Intercept, Slope, dan R-Square

```python
a = regressor.intercept_
b = regressor.coef_

print("a. Estimated model intercept, a:", a)
print("b. Estimated model slope, b:", b)
for i, col in enumerate(df.columns[1:-1]):
    print(f"   Slope b{i+1} ({col}): {b[i]}")
```

```python
from sklearn.metrics import r2_score
score = r2_score(y_test, y_pred)
print("c. R-Square:", score)
```

## 1. Membaca file konsumen.csv

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('konsumen.csv', sep=';')
print("Dimensi dataset konsumen (baris, kolom):", df.shape)
```

### 2. Menampilkan sebanyak 7 row data teratas

```python
df.head(7)
```

### 3. Menghapus kolom No

```python
df = df.drop(columns=['No'])
print("Kolom 'No' berhasil dihapus. Dimensi data sekarang:", df.shape)
```

### 4. Menampilkan 9 row data teratas

```python
df.head(9)
```

### 5. Menampilkan grafik hubungan gaji dan pengeluaran

```python
plt.scatter(df['Gaji'], df['Pengeluaran'], color='blue')
plt.title('Grafik Hubungan Gaji dan Pengeluaran')
plt.xlabel('Gaji')
plt.ylabel('Pengeluaran')
plt.show()
```

### 6. Gunakan 2 klaster

```python
from sklearn.cluster import KMeans

X = df[['Gaji', 'Pengeluaran']].values

kmeans = KMeans(n_clusters=2, random_state=0)
kmeans.fit(X)

print("Hasil label klaster tiap konsumen:\n", kmeans.labels_)
```

### 7. Menampilkan nilai pusat klaster

```python
pusat_klaster = kmeans.cluster_centers_
print("Nilai pusat klaster (Centroid):\n", pusat_klaster)

for i, center in enumerate(pusat_klaster):
    print(f"Klaster {i}: Gaji rata-rata = {center[0]:.2f}, Pengeluaran rata-rata = {center[1]:.2f}")
```

### 8. Menampilkan Grafik Konsumen yang memiliki 2 warna berbeda

```python
plt.scatter(X[kmeans.labels_ == 0, 0], X[kmeans.labels_ == 0, 1], color='red', label='Klaster 0')
plt.scatter(X[kmeans.labels_ == 1, 0], X[kmeans.labels_ == 1, 1], color='blue', label='Klaster 1')
plt.title('Grafik Konsumen (2 Klaster Berwarna Berbeda)')
plt.xlabel('Gaji')
plt.ylabel('Pengeluaran')
plt.legend()
plt.show()
```

### 9. Menampilkan Grafik Konsumen yang memiliki 2 warna berbeda dengan pusat klaster

```python
plt.scatter(X[kmeans.labels_ == 0, 0], X[kmeans.labels_ == 0, 1], color='red', label='Klaster 0')
plt.scatter(X[kmeans.labels_ == 1, 0], X[kmeans.labels_ == 1, 1], color='blue', label='Klaster 1')
plt.scatter(pusat_klaster[:, 0], pusat_klaster[:, 1], s=250, color='black', marker='X', edgecolors='white', label='Pusat Klaster (Centroid)')
plt.title('Grafik Konsumen (2 Klaster dengan Pusat Klaster)')
plt.xlabel('Gaji')
plt.ylabel('Pengeluaran')
plt.legend()
plt.show()
```
