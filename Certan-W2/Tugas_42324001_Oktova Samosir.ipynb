# Analisis Customer Shopping Trends

Pada tugas ini, kita akan melakukan analisis terhadap dataset

## Variabel yang Digunakan

Sesuai dengan soal, analisis korelasi dilakukan terhadap tiga variabel,
yaitu `temperature_fahrenheit`, `pressure_mb`, dan
`air_quality_Sulphur_dioxide`.


```python
tabel_variabel_weather = weather_df[
    [
        'temperature_fahrenheit',
        'pressure_mb',
        'air_quality_Sulphur_dioxide'
    ]
]

tabel_variabel_weather
```

```python
tabel_korelasi_weather = pd.DataFrame({
    'Variabel 1': [
        'temperature_fahrenheit',
        'temperature_fahrenheit',
        'pressure_mb'
    ],
    'Variabel 2': [
        'pressure_mb',
        'air_quality_Sulphur_dioxide',
        'air_quality_Sulphur_dioxide'
    ],
    'Nilai Korelasi': [
        weather_df['temperature_fahrenheit'].corr(
            weather_df['pressure_mb']
        ),
        weather_df['temperature_fahrenheit'].corr(
            weather_df['air_quality_Sulphur_dioxide']
        ),
        weather_df['pressure_mb'].corr(
            weather_df['air_quality_Sulphur_dioxide']
        )
    ]
})

tabel_korelasi_weather
```

## Interpretasi Hasil Korelasi

Berdasarkan hasil analisis korelasi, hubungan antara
`temperature_fahrenheit` dan `pressure_mb` memiliki nilai korelasi
sebesar **-0.486833**. Nilai tersebut menunjukkan hubungan negatif
dengan kekuatan sedang. Artinya, ketika `temperature_fahrenheit`
meningkat, `pressure_mb` cenderung menurun.

Hubungan antara `temperature_fahrenheit` dan
`air_quality_Sulphur_dioxide` memiliki nilai korelasi sebesar
**-0.021532**. Nilai tersebut mendekati 0 sehingga menunjukkan
hubungan negatif yang sangat lemah.

Hubungan antara `pressure_mb` dan
`air_quality_Sulphur_dioxide` memiliki nilai korelasi sebesar
**0.018620**. Nilai tersebut juga sangat dekat dengan 0 sehingga
menunjukkan hubungan positif yang sangat lemah.

Dari ketiga pasangan variabel tersebut, hubungan paling kuat
terdapat antara `temperature_fahrenheit` dan `pressure_mb`,
sedangkan hubungan paling lemah terdapat antara `pressure_mb`
dan `air_quality_Sulphur_dioxide`.

## Analisis Korelasi

Korelasi digunakan untuk mengetahui hubungan linear antara

## Analisis Korelasi Purchase Amount (USD) vs Review Rating

Pada tahap ini dilakukan analisis korelasi antara
`Purchase Amount (USD)` dan `Review Rating`.

Analisis ini bertujuan untuk mengetahui arah dan kekuatan hubungan
linear antara jumlah pembelian dengan rating yang diberikan oleh
pelanggan.

```python
corr_purchase_rating = df['Purchase Amount (USD)'].corr(
    df['Review Rating']
)

tabel_corr_purchase_rating = pd.DataFrame({
    'Variabel 1': ['Purchase Amount (USD)'],
    'Variabel 2': ['Review Rating'],
    'Nilai Korelasi': [corr_purchase_rating]
})

tabel_corr_purchase_rating
```

## Interpretasi Korelasi Purchase Amount (USD) vs Review Rating

Berdasarkan tabel, nilai korelasi antara `Purchase Amount (USD)` dan
`Review Rating` adalah 0.030776.

Nilai tersebut bernilai positif tetapi sangat dekat dengan 0.
Hal ini menunjukkan bahwa hubungan linear antara jumlah pembelian
dan rating pelanggan bersifat positif dan sangat lemah.

Dengan demikian, berdasarkan dataset yang digunakan,
`Purchase Amount (USD)` tidak menunjukkan hubungan linear yang kuat
dengan `Review Rating`.
