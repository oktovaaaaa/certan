## Analisis Korelasi

Korelasi digunakan untuk mengetahui hubungan antara dua atau lebih
variabel.

Nilai korelasi berada pada rentang -1 sampai 1:

1 =  korelasi positif.
1 = korelasi negatif.
0 = tidak terdapat korelasi.

Fungsi `.corr()` digunakan untuk menghitung nilai korelasi antar
kolom tersebut.
