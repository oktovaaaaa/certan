```python
import numpy as np
# Identifikasi Q1 Q3
q1, q3 = np.percentile(df2['Height'], [25, 75])
# Hitung IQR
iqr = q3 - q1
# Cari lower dan upper bounds
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)
# Hapus outliers
clean_data = df2[(df2['Height'] >= lower_bound)
 & (df2['Height'] <= upper_bound)]
sns.boxplot(clean_data['Height'])
```

Kode ini digunakan untuk menghitung batas outlier menggunakan Q1, Q3, dan IQR, lalu menghapus data outlier dan menampilkan boxplot setelah data dibersihkan.

```python
import matplotlib.pyplot as plt
# Membuat daftar kolom yang ingin dianalisis, yaitu "Weight" dan "Height".
columns_to_plot = ['Weight', 'Height']
# Memastikan kolom ada di dataset
columns_to_plot = [col for col in columns_to_plot if col in df2.columns]
fig, axs = plt.subplots(len(columns_to_plot), 1, dpi=95, figsize=(7, 5 * len(columns_to_plot)))
if len(columns_to_plot) == 1:
 axs = [axs]
for i, col in enumerate(columns_to_plot):
 axs[i].boxplot(df2[col], vert=False)
 axs[i].set_ylabel(col)
plt.show()
```

Kode ini digunakan untuk membuat boxplot dari kolom Weight dan Height secara bersamaan agar outlier pada kedua kolom dapat dilihat dengan mudah.


```python
import matplotlib.pyplot as plt
import numpy as np


# Fungsi untuk menghapus outliers secara berurutan
def remove_outliers_sequentially(df2, columns):
  df_cleaned = df2.copy()
  for col in columns:
    q1, q3 = np.percentile(df_cleaned[col], [25, 75])
    iqr = q3 - q1
    lower_bound = q1 - (1.5 * iqr)
    upper_bound = q3 + (1.5 * iqr)
    df_cleaned = df_cleaned[
        (df_cleaned[col] >= lower_bound) & (df_cleaned[col] <= upper_bound)
    ]
  return df_cleaned


# Terapkan fungsi untuk menghapus outliers dari "Height" dan "Weight"
clean_data = remove_outliers_sequentially(df2, ['Height', 'Weight'])

# Plot boxplots setelah menghapus outliers
columns_to_plot = ['Weight', 'Height']
fig, axs = plt.subplots(
    len(columns_to_plot), 1, dpi=95, figsize=(7, 5 * len(columns_to_plot))
)

if len(columns_to_plot) == 1:
  axs = [axs]

for i, col in enumerate(columns_to_plot):
  axs[i].boxplot(clean_data[col], vert=False)
  axs[i].set_ylabel(col)

plt.show()
```

Kode ini digunakan untuk menghapus outlier pada kolom Height dan Weight secara berurutan menggunakan metode IQR, kemudian menampilkan boxplot setelah data dibersihkan.


```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler(feature_range=(0, 1))
# Memisahkan kolom numerik dan non-numerik
numeric_cols = clean_data.select_dtypes(include=[np.number]).columns # Ambil kolom
non_numeric_cols = clean_data.select_dtypes(exclude=[np.number]).columns # Ambil k
rescaled_numeric = scaler.fit_transform(clean_data[numeric_cols])
rescaled_numeric[:]
```

## TUGAS 3: Handling Outliers dari Dataset 'diabetes.csv

## Handling Outliers - Diabetes Dataset
Membaca dataset diabetes dan memeriksa data awal. Tiga kolom yang dipilih untuk deteksi outliers adalah 'Glucose', 'BloodPressure', dan 'BMI'.

```python
import matplotlib.pyplot as plt
import numpy as np

path_diabetes = "diabetes.csv"
df_diabetes = pd.read_csv(path_diabetes)
df_diabetes.head()
```

Mendeteksi outliers secara visual menggunakan boxplot dari matplotlib untuk ketiga kolom yang dipilih sebelum dilakukan pembersihan data.

```python
# Membuat daftar kolom yang ingin dianalisis
columns_to_plot = ['Glucose', 'BloodPressure', 'BMI']
columns_to_plot = [col for col in columns_to_plot if col in df_diabetes.columns]

fig, axs = plt.subplots(len(columns_to_plot), 1, dpi=95, figsize=(7, 4 * len(columns_to_plot)))

if len(columns_to_plot) == 1:
    axs = [axs]

for i, col in enumerate(columns_to_plot):
    axs[i].boxplot(df_diabetes[col], vert=False)
    axs[i].set_ylabel(col)
    axs[i].set_title(f'Sebelum Outlier Dihapus - {col}')

plt.tight_layout()
plt.show()
```

Mendefinisikan fungsi untuk menghapus outliers secara berurutan pada kolom-kolom yang ditentukan menggunakan metode Interquartile Range (IQR).

```python
def remove_outliers_sequentially(df, columns):
    df_cleaned = df.copy()
    for col in columns:
        q1, q3 = np.percentile(df_cleaned[col], [25, 75])
        iqr = q3 - q1
        lower_bound = q1 - (1.5 * iqr)
        upper_bound = q3 + (1.5 * iqr)
        df_cleaned = df_cleaned[(df_cleaned[col] >= lower_bound) & (df_cleaned[col] <= upper_bound)]
    return df_cleaned

# Terapkan fungsi untuk menghapus outliers pada kolom yang dipilih
clean_diabetes = remove_outliers_sequentially(df_diabetes, columns_to_plot)
```

Membuat boxplot kembali setelah outliers dihapus untuk memverifikasi bahwa titik data di luar whiskers sudah berkurang / hilang.

```python
fig, axs = plt.subplots(len(columns_to_plot), 1, dpi=95, figsize=(7, 4 * len(columns_to_plot)))

if len(columns_to_plot) == 1:
    axs = [axs]

for i, col in enumerate(columns_to_plot):
    axs[i].boxplot(clean_diabetes[col], vert=False)
    axs[i].set_ylabel(col)
    axs[i].set_title(f'Setelah Outlier Dihapus - {col}')

plt.tight_layout()
plt.show()
```

Memeriksa perbandingan ukuran dataset sebelum dan sesudah tahap pembersihan outliers.

```python
print("Ukuran dataset sebelum dibersihkan:", df_diabetes.shape)
print("Ukuran dataset sesudah dibersihkan:", clean_diabetes.shape)
```
