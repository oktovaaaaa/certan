```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

# load datasdfet iris darisickit learn
iris = load_iris()
```


### Penjelasan:
Code ini dipakai untuk menyiapkan alat dan data. import numpy dan import pandas untuk mengolah array dan tabel, lalu from sklearn.datasets import ... untuk mengambil fungsi load_iris. 
 iris = load_iris() untuk memuat dataset iris.

```python
# konfersi data menjadi dataframe pandas 
df =   pd.DataFrame(data=iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

#def kan x(fitur) dan y(label/target)
X = df.drop(columns=["Species"])
y = df["Species"]

print(X)
```

### Penjelasan:

Code ini dipakai untuk mengubah data iris menjadi tabel dan memisahkan fitur dengan labelnya. df = pd.dataframe(...) untuk membuat tabel dari data iris, dengan columns=... sebagai nama kolomnya. df["species"] = ... untuk menambah kolom species yang isinya nama spesies bunga.

x = df.drop(...) untuk mengambil semua kolom kecuali species sebagai fitur, dan y = df["species] untuk mengambil kolom species sebagai label. print(x) untuk menampilkan 150 baris dan 4 kolom.

```python
y
```

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

### Penjelasan:
Code ini dipakai untuk mengubah label y dari teks menjadi angka. from sklearn.preprocessing import ... untuk mengambil fungsi labelencoder, lalu lab = labelencoder() untuk membuat objek encodernya. y = lab.fit_transform(y) untuk mengubah isi y jadi angka sesuai urutan kelasnya. y untuk menampilkan hasilnya.

```python
from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

### Penjelasan:

Code ini dipakai untuk membagi data menjadi data latih dan data uji. from sklearn.model_selection import ... untuk mengambil fungsi train_test_split. x_train, x_test = train_test_split(x, ...) untuk membagi fitur, yang dimana test_size=0.2 artinya 20% data jadi data uji. y_train, y_test = train_test_split(y, ...) untuk membagi label dengan cara yang sama, dan random_state=0 dipakai supaya pembagian acaknya sama antara x dan y.

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

### Penjelasan:
Code ini dipakai untuk membuat dan melatih model naive bayes. from sklearn.naive_bayes import ... untuk mengambil fungsi gaussiannb. model = gaussiannb() untuk membuat objek modelnya, dan nb_model = model.fit(x_train, y_train) untuk melatih model itu dengan data latih.

```python
result = NB_model.predict(X_test)
result
```

### Penjelasan:

Code ini dipakai untuk memprediksi kelas dari data uji. result untuk menyimpan hasil prediksi dari nb_model terhadap x_test. result untuk menampilkan hasilnya.

```python
from sklearn.metrics import accuracy_score
print("akurasi of data test: ", accuracy_score(y_test, result)*100)
```

### Penjelasan:
Code ini dipakai untuk menghitung akurasi model. from sklearn.metrics import ... untuk mengambil fungsi accuracy_score dan menampilkan akurasi, yang dihitung dengan membandingkan y_test dan result, lalu dikali 100 supaya jadi persen.
