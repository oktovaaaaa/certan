## Tugas 1

```python
#1
import pandas as pd
df = pd.read_csv('rounded_hours_student_scores.csv')
print(df.head())
```

```python
# 2
X = df.iloc[:, :-1].values
y = df.iloc[:, 1].values
```

```python
# 3
import matplotlib.pyplot as plt
plt.scatter(X, y, color = "red")
plt.xlabel('Hours Studied')
plt.ylabel('Scores')
plt.show()
```

```python
# 4
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

```python
# 5
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
# 6
a = regressor.intercept_
b = regressor.coef_[0]

print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)
print(f"ŷ = {a:.2f} + {b:.2f}x")
```

```python
# 7
y_pred = regressor.predict(X_test)
y_pred
```

```python
# 8
from sklearn.metrics import r2_score
score = r2_score(y_test, y_pred)
score
```

```python
# 9
plt.scatter(X_train, y_train, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
plt.title('Scores vs Hours Studied (Training set)')
plt.xlabel('Hours Studied')
plt.ylabel('Scores')
plt.show()
```

```python
# 10
plt.scatter(X_test, y_test, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
plt.title('Scores vs Hours Studied (Test set)')
plt.xlabel('Hours Studied')
plt.ylabel('Scores')
plt.show()
```
